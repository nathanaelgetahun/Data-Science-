```python
df = pd.read_csv("futball.csv")
display(df)
```

```python
display(df.isnull().sum())   #displays numbers of missing values for each column
df.duplicated()
```

```python
df["Pass_Accuracy"] = df["Passes_Completed"] / df["Passes_Attempted"] * 100
df["Goals_Per_90"] = df["Goals"] / df["Minutes"] * 90
df["Goal_Contributions"] = df["Goals"] + df["Assists"]

display(df)
```

```python
display(df[df["Goals"] >= 15])
display(df[df["Pass_Accuracy"] > 80])
display(df[(df["Position"] == "Forward") & (df["Minutes"] > 2600)])
```

```python
#have to set to a new variable because groupby makes new object
positions_goals= df.groupby("Position")
display(positions_goals["Goals"].mean())
```

```python
teams_goals = df.groupby(["Team"])
display(teams_goals["Goals"].sum())
```

```python
#you have to save this df cause the old version doesn't have pass accuracy
df.to_csv("futball.csv", index=False)
team_pass_accuracy = df.groupby(["Team"])
display(team_pass_accuracy["Pass_Accuracy"].mean())
```
